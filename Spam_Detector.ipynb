# Spam SMS Classifier

## 1. Setup and Data Loading
This cell loads necessary libraries, reads the SMS spam dataset, and preprocesses the labels.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Load data - grab sms_spam.csv from Kaggle
df = pd.read_csv('spam.csv', encoding='latin-1')[['v1', 'v2']]
df.columns = ['label', 'text']

# Convert labels: ham=0, spam=1
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

print("Data loaded and labels converted. First 5 rows:")
display(df.head())
```

## 2. Data Splitting and Vectorization
This cell splits the data into training and testing sets and converts text into numerical features using `CountVectorizer`.

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.2, random_state=42
)
print(f"Data split into {len(X_train)} training samples and {len(X_test)} testing samples.")

# Convert text to word counts
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
print("Text converted to word count vectors using CountVectorizer.")
```

## 3. Model Training
Here, a Multinomial Naive Bayes model is initialized and trained on the vectorized training data.

```python
# Train model
model = MultinomialNB()
model.fit(X_train_vec, y_train)
print("Multinomial Naive Bayes model trained successfully.")
```

## 4. Model Evaluation
This cell evaluates the trained model's accuracy on the test set.

```python
# Test model
preds = model.predict(X_test_vec)
accuracy = accuracy_score(y_test, preds)
print(f"Model Accuracy on the test set: {accuracy:.2f}")
```

## 5. Spam Prediction Function
This function takes a text message as input and uses the trained model to predict if it's spam or not, along with the spam probability.

```python
def predict_spam(text):
    text_vec = vectorizer.transform([text])
    prob = model.predict_proba(text_vec)[0][1]
    result = "SPAM!!!" if prob > 0.5 else "NOT SPAM."
    print(f"'{text}' → {result} | Spam chance: {prob:.0%}")
```

## 6. Interactive Spam Messenger
Use the input box below to type a message and see if the model predicts it as spam. Type 'exit' to stop.

```python
# Interactive Spam Predictor
print("Enter text to check for spam. Type 'exit' to quit.\n")
while True:
    user_input = input("Your message: ")
    if user_input.lower() == 'exit':
        break
    predict_spam(user_input)
print("\nExiting interactive spam predictor.")
```
